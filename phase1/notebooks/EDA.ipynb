Categorical risk analysis

```python


categorical_features = [
    "merchant_category",
    "foreign_transaction",
    "location_mismatch"
]

for feature in categorical_features:

    print(f"\nFraud analysis for: {feature}")

    fraud_rate = (
        df.groupby(feature)["is_fraud"]
        .mean()
        .mul(100)
        .sort_values(ascending=False)
    )

    print("Fraud rate (%):")
    display(
        fraud_rate.round(2).to_frame("Fraud Rate (%)")
    )

    plt.figure(figsize=(8, 4))

    fraud_rate.plot(kind="bar")

    plt.title(f"Fraud Rate by {feature}")
    plt.xlabel(feature)
    plt.ylabel("Fraud Rate (%)")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()
```

Chi-Square Test

```python
from scipy.stats import chi2_contingency

chi_square_results = []

chi_features = [
    "merchant_category",
    "foreign_transaction",
    "location_mismatch"
]

chi_features = [
    col for col in chi_features
    if col in df.columns
]

for feature in chi_features:

    contingency_table = pd.crosstab(
        df[feature],
        df["is_fraud"]
    )

    chi2, p_value, dof, expected = chi2_contingency(
        contingency_table
    )

    chi_square_results.append({
        "Feature": feature,
        "Chi2": chi2,
        "p_value": p_value,
        "Significant": p_value < 0.05
    })

chi_square_df = pd.DataFrame(
    chi_square_results
)

print("CHI-SQUARE TEST RESULTS")

display(
    chi_square_df.sort_values(
        "p_value"
    )
)
```

ANOVA

```python
from scipy.stats import f_oneway

# Numerical columns to test
anova_features = [
    "amount",
    "transaction_hour",
    "device_trust_score",
    "velocity_last_24h"
]

anova_results = []

for feature in anova_features:

    non_fraud = df[df["is_fraud"] == 0][feature].dropna()
    fraud = df[df["is_fraud"] == 1][feature].dropna()

    f_statistic, p_value = f_oneway(
        non_fraud,
        fraud
    )

    anova_results.append({
        "Feature": feature,
        "F_Statistic": f_statistic,
        "p_value": p_value,
        "Significant": p_value < 0.05
    })

anova_df = pd.DataFrame(anova_results)

print("ANOVA TEST RESULTS")
display(anova_df.sort_values("p_value"))
```

Pearson Correlation Significance

```python
from scipy.stats import pearsonr

# Numerical features to test
correlation_features = [
    "amount",
    "transaction_hour",
    "device_trust_score",
    "velocity_last_24h"
]

correlation_results = []

for feature in correlation_features:

    temp = df[[feature, "is_fraud"]].dropna()

    correlation, p_value = pearsonr(
        temp[feature],
        temp["is_fraud"]
    )

    correlation_results.append({
        "Feature": feature,
        "Correlation": correlation,
        "p_value": p_value,
        "Significant": p_value < 0.05
    })

correlation_significance_df = pd.DataFrame(correlation_results)

print("PEARSON CORRELATION SIGNIFICANCE RESULTS")

display(
    correlation_significance_df.sort_values("p_value")
)
```

FINAL STATISTICALLY JUSTIFIED FEATURE SHORTLIST

```python
chi_significant = chi_square_df[
    chi_square_df["p_value"] < 0.05
]["Feature"].tolist()

anova_significant = anova_df[
    anova_df["p_value"] < 0.05
]["Feature"].tolist()

correlation_significant = correlation_significance_df[
    correlation_significance_df["p_value"] < 0.05
]["Feature"].tolist()

print("Significant Chi-Square Features:")
print(chi_significant)

print("\nSignificant ANOVA Features:")
print(anova_significant)

print("\nSignificant Correlation Features:")
print(correlation_significant)
```

Final Summary

```python
feature_set = set(
    chi_significant +
    anova_significant +
    correlation_significant
)

final_features = sorted(feature_set)

print("\nFINAL STATISTICAL FEATURE SHORTLIST")
print("=" * 45)

for feature in final_features:
    print("✓", feature)
```

conclusive table

```python
feature_summary = pd.DataFrame({
    "Feature": sorted(
        set(
            chi_significant +
            anova_significant +
            correlation_significant
        )
    )
})

feature_summary["Chi_Square"] = (
    feature_summary["Feature"]
    .isin(chi_significant)
)

feature_summary["ANOVA"] = (
    feature_summary["Feature"]
    .isin(anova_significant)
)

feature_summary["Correlation_Significant"] = (
    feature_summary["Feature"]
    .isin(correlation_significant)
)

feature_summary["Number_of_Significant_Tests"] = (
    feature_summary[
        [
            "Chi_Square",
            "ANOVA",
            "Correlation_Significant"
        ]
    ].sum(axis=1)
)

feature_summary = feature_summary.sort_values(
    "Number_of_Significant_Tests",
    ascending=False
)

display(feature_summary)
```

Final Conclusion

# Phase 1 — Statistical Feature Selection Conclusion

The exploratory analysis was extended beyond descriptive statistics
using formal statistical hypothesis tests.

### Chi-Square Test
Chi-square tests were used to identify statistically significant
associations between categorical variables and fraudulent
transactions.

### ANOVA
One-way ANOVA was performed to determine whether numerical feature
means differed significantly between fraudulent and non-fraudulent
transactions.

### Correlation Significance
Pearson correlation significance testing was performed to determine
whether numerical features had statistically significant relationships
with the fraud target.

### Feature Shortlisting
Features with p-values below 0.05 were considered statistically
significant and shortlisted for subsequent machine-learning analysis.

Statistical significance is used as one feature-selection criterion.
Final feature selection will also consider model performance,
multicollinearity, business relevance, and data leakage.
